```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sort the DataFrame by Year for a proper time sequence
df = df.sort_values(by='Year')

# Melt the DataFrame to a "long" format
df_melted = df.melt(
    id_vars='Year', 
    value_vars=['Total_Revenue_Impact', 'High_Risk_Cases'], 
    var_name='Metric', 
    value_name='Value'
)

plt.figure(figsize=(10,6))
sns.lineplot(data=df_melted, x='Year', y='Value', hue='Metric', marker='o')
plt.title('Trends Over Time: Revenue Impact vs. High Risk Cases')
plt.xlabel('Year')
plt.ylabel('Value')
plt.legend(title='Metric')
plt.show()
```

# Overall Insights & Conclusions

## High Risk Cases Over Time
The line chart shows that the number of high risk cases fluctuates between about 200 and 450 over the 2020–2025 period. 

- Deloitte experiences a sharp spike in 2023, reaching nearly 450 cases, which indicates a significant surge in risk exposure that year.
- PwC remains relatively stable throughout the period, with numbers generally ranging between 350 and 400 cases.
- KPMG and Ernst & Young show moderate variability, with their figures mostly falling between 250 and 400 cases.

These differences may be due to varying risk management practices, differences in client profiles, or external factors that affect each firm differently.

## Compliance Violations and Fraud Cases
When comparing firms, the bar chart for compliance violations indicates that Ernst & Young has the highest average—around 300 violations—while PwC has the lowest, with approximately 200 violations. In addition, the relationship between fraud cases detected and total audit engagements is moderate, with a correlation of about 0.27. This suggests that while increased audit activity may uncover more fraud cases, the relationship is not very strong.

This might mean that differences in internal controls and the quality of audits are influencing these variations across the firms.

## Total Revenue Impact
The trend line for total revenue impact shows a gentle upward trajectory from 2020 to 2025. This could be interpreted as a sign of gradual improvements in market performance or successful client retention strategies. However, there isn’t a strong direct link between revenue impact and risk metrics like high risk cases or fraud cases. This disconnect suggests that revenue performance is likely influenced by multiple factors beyond the risk indicators observed.

## Correlation Analysis and Combined Insights
The correlation matrix and heatmap reveal that most variables show low to moderate correlations. The strongest relationship is between total audit engagements and fraud cases detected (around 0.27). This reinforces the idea that more audits may reveal more instances of fraud, though only modestly.

These findings indicate that while some risk metrics tend to move together, each metric is also influenced by its own set of factors. A deeper, non-linear analysis might be necessary to uncover more detailed insights.

## Key Takeaways and Future Recommendations
- Deloitte’s spike in high risk cases in 2023 suggests significant internal or external pressures that year.
- PwC’s steady numbers imply a robust and consistent approach to risk management.
- For a more complete picture, future analysis should incorporate external data, such as economic indicators, and explore non-linear models to capture the complex relationship between risk factors and revenue performance.

Overall, this analysis provides a snapshot of risk and compliance trends among the Big 4 firms. While the observed patterns offer insight into each firm’s risk profile, further investigation with additional data would help build a more comprehensive understanding.
